```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("Titanic.csv")
```

```python
df.head()
```

# **Categorical Data**

## **Count Plot**

```python
sns.countplot(data=df,x='Survived',palette='Set2')
df['Survived'].value_counts()
```

```python
sns.countplot(data=df,x='Pclass',palette='Set2')
df['Survived'].value_counts()
```

```python
sns.countplot(data=df,x='Sex',palette='Set2')
df['Sex'].value_counts()
```

```python
sns.countplot(data=df,x='Embarked',palette='Set2')
df['Embarked'].value_counts()
```

## **Pie Chart**

```python
## if we want to represent the same data in term of percentage
df['Survived'].value_counts().plot(kind="pie",autopct="%.2f") #autopct shows percentage
```

```python
df['Pclass'].value_counts().plot(kind="pie",autopct="%.2f")
```

```python
df['Sex'].value_counts().plot(kind="pie",autopct="%.2f")
```

# **Numerical Data**

```python
sns.displot(data=df,x='Age',kde=True)
```
