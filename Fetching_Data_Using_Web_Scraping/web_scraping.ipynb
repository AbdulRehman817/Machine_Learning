```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
soup=BeautifulSoup(webpage,'lxml')
for i in soup.find_all('h3'):
    print(i.a['title']) # title was inside a tag so first we went using i.a and then title
```

```python
for i in soup.find_all("p",class_='price_color'):
    print(i.text)
```

```python
for i in soup.find_all("p",class_="instock availability"):
    print(i.text.strip())
```

```python

books = soup.find_all("article", class_="product_pod")
```

```python

rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}

title=[]
price=[]
instock=[]
ratings=[]
for i in books:
    tit=i.find('h3').a['title']
    pri=i.find('p',class_='price_color').text
    ins=i.find('p',class_='instock availability').text.strip()
    word=i.find('p',class_='star-rating')['class'][1]
    title.append(tit)
    instock.append(ins)
    price.append(pri)
    ratings.append(rating_map[word])
p={"title":title,"price":price,"ratings":ratings,"instock":instock}
pd.DataFrame(p)
    
```
