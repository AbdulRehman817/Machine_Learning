# Opening a csv file from an URL

```python
from io import StringIO
import pandas as pd
import requests

url = "https://gist.githubusercontent.com/kevin336/acbb2271e66c10a5b73aacf82ca82784/raw/e38afe62e088394d61ed30884dd50a6826eee0a8/employees.csv"

# Add a standard browser User-Agent header
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

# Fetch the page content
response = requests.get(url, headers=headers)

# Wrap the text response and read it into pandas
df = pd.read_csv(StringIO(response.text))
```

# Separate Parameters

```python
df=pd.read_csv("movie_titles.tsv")# names is used when we donot have proper column names and we have to add column name by ourself
df.head()
```

```python
df=pd.read_csv("movie_titles.tsv",sep='\t')
df.head()
```

```python
df=pd.read_csv("movie_titles.tsv",sep='\t',names=['sno','name','release_year','votes','rating','genre'])# names is used when we donot have proper column names and we have to add column name by ourself
df.head()
```

# Index Col Parameter

```python
df=pd.read_csv("aug_test.csv") #index_col is used when we want our column to come in index instead of pandas generated index
df.head()
```

```python
df=pd.read_csv("aug_test.csv",index_col=['enrollee_id']) #index_col is used when we want our column to come in index instead of pandas generated index
df.head()
```
